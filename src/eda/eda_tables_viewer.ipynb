```python

TABLES_B64 = """"""
```

```python
import base64, gzip, json
from io import StringIO
from IPython.display import display, Markdown
import pandas as pd

_tables = json.loads(gzip.decompress(base64.b64decode(TABLES_B64)).decode("utf-8"))
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
pd.set_option("display.width", 180)

for _name, _content in _tables.items():
    display(Markdown(f"### `{_name}`"))
    if _name.endswith(".json"):
        _obj = json.loads(_content)
        _df = pd.DataFrame([{"metric": k, "value": json.dumps(v, ensure_ascii=False) if isinstance(v, (dict, list)) else v} for k, v in _obj.items()])
    else:
        _df = pd.read_csv(StringIO(_content))
    display(Markdown(f"{_df.shape[0]:,} dòng × {_df.shape[1]:,} cột"))
    if len(_df) > 100:
        display(_df.head(20))
        display(Markdown("Bảng lớn — phần trên là 20 dòng đầu tiên."))
    else:
        display(_df)
```
